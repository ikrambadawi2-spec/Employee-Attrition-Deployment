
### Dataset Description

| Column                   | Description                                                                       | 
| ------------------------ | --------------------------------------------------------------------------------- | 
| `Enrollee_id`            | Unique ID assigned to each candidate.                                             | 
| `City`                   | The city where the candidate lives.                                               |            
| `City_development_index` | Development index of the candidate's city.                                        |           
| `Gender`                 | Gender of the candidate.                                                          |            
| `Relevent_experience`    | Indicates whether the candidate has relevant work experience.                     |            
| `Enrolled_university`    | University enrollment status of the candidate.                                    |            
| `Education_level`        | Highest education level attained by the candidate (Graduate, Masters, PhD, etc.). |            
| `Major_discipline`       | Candidate's main field of study (STEM, Business, Arts, Humanities, etc.).         |            
| `Experience`             | Number of years of work experience.                                               |   
| `Company_size`           | Number of employees in the candidate's current or previous company.               |   
| `Company_type`           | Type of company where the candidate currently or previously worked.               |            
| `Last_new_job`           | Number of years since the candidate's last job change.                            |   
| `Training_hours`         | Number of hours of training completed by the candidate.                           |           
| `Target`                 | Indicates whether the candidate is looking for a new job (1 = Yes, 0 = No).       |

# Feature Engineering

```python
def career_stage(exp):
    if exp == 0:
        return 'Fresher'
    elif exp <= 2:
        return 'Junior'
    elif exp <= 5:
        return 'Mid-Level'
    elif exp <= 10:
        return 'Senior'
    else:
        return 'Expert'

df['Career_Stage'] = df['Experience'].apply(career_stage)
df['Career_Stage']
```

```python
df.head()
```

- Split Data into Input Features and Target Feature

```python
x = df.drop('Target', axis= 1)
y = df['Target']
```

### Num Pipeline 1

```python
# ( City_development_index, Training_hours) Scale with "RobustScaler":

from sklearn.pipeline import Pipeline

from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

num_pipeline_1 = Pipeline(steps= [ ('RobustScaler', scaler) ])

num_pipeline_1
```

```python
num_cols2 = ['Experience']
num_cols2
```

### Cat Pipeline 1

```python
cat_cols_1 = ['Gender', 'Enrolled_university', 'Last_new_job']
```

```python
# (Gender, Enrolled_university, Last_new_job) columns impute with 'mode' & encode with 'onehot encoder'

from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

mode_imputer = SimpleImputer(strategy='most_frequent')

ohe = OneHotEncoder(drop= 'first', sparse_output= False)

cat_pipeline_1 = Pipeline(steps= [ ('Mode Imputer', mode_imputer),
                                   ('OHE', ohe) ])
cat_pipeline_1
```

### Cat Pipeline 2

```python
cat_cols_2 = ['Major_discipline', 'Company_type']
```

```python
# (Major_discipline, Company_type) ---> impute with 'Other' & encode with 'onehot encoder'

from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

const_imputer = SimpleImputer(strategy= 'constant', fill_value= 'Other')

ohe = OneHotEncoder(drop= 'first', sparse_output= False)

cat_pipeline_2 = Pipeline(steps= [ ('Const Imputer', const_imputer),
                                   ('OHE', ohe) ])
cat_pipeline_2
```

### Cat Pipeline 3

```python
cat_col_3 = ['Relevent_experience']
```

```python
# 'Relevent_experience'---> encode with 'onehot encoder

from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop= 'first', sparse_output= False)

cat_pipeline_3 = Pipeline(steps= [ ('OHE', ohe) ])
cat_pipeline_3
```

### Cat Pipeline 4

```python
cat_col_4 = ['City']
```

```python
# City ---> Encode with Binary Encoder

from category_encoders import BinaryEncoder

BE = BinaryEncoder()

cat_pipeline_4 = Pipeline(steps= [ ('BE', BE)])
cat_pipeline_4
```

### Cat Pipeline 5

```python
ordinal_cols1 = ['Education_level', 'Company_size']
```

```python
x['Education_level'].unique()
```

```python
x['Company_size'].unique()
```

```python
# ('Education_level'and 'Company_size') impute with "most frequent" & encode by ordinal encoder

from sklearn.preprocessing import OrdinalEncoder

mode_imputer = SimpleImputer(strategy='most_frequent')

ord_encoder = OrdinalEncoder(categories= [ ['Primary School', 'High School', 'Graduate', 'Masters', 'Phd'], 
                                           ['<10', '10-49', '50-99', '100-500', '500-999', '1000-4999', '5000-9999', '10000+'] ] )

cat_pipeline_5 = Pipeline(steps= [ ('Mode Imputer', mode_imputer),
                                   ('Ord Encoder', ord_encoder) ])
cat_pipeline_5
```

### Cat Pipeline 6

```python
ordinal_cols2 = ['Career_Stage']
```

```python
df['Career_Stage'].unique()
```

```python
# ( Career_Stage)---> encode by ordinal encoder

from sklearn.preprocessing import OrdinalEncoder

ord_encoder = OrdinalEncoder(categories= [ ['Fresher', 'Junior', 'Mid-Level', 'Senior', 'Expert'] ] )

cat_pipeline_6 = Pipeline(steps= [('Ord Encoder', ord_encoder) ])
cat_pipeline_6
```

### Assign each column to its corrsponding Pipeline

```python
from sklearn.compose import ColumnTransformer

preprocessing = ColumnTransformer(transformers= [
    ('Num Pipeline 1 ', num_pipeline_1, num_cols1),
    ('Num Pipeline 2 ', num_pipeline_2, num_cols2),
    ('Cat Pipeline 1 ', cat_pipeline_1, cat_cols_1),
    ('Cat Pipeline 2 ', cat_pipeline_2, cat_cols_2),
    ('Cat Pipeline 3 ', cat_pipeline_3, cat_col_3),
    ('Cat Pipeline 4 ', cat_pipeline_4, cat_col_4),
    ('Cat Pipeline 5 ', cat_pipeline_5, ordinal_cols1),
    ('Cat Pipeline 6 ', cat_pipeline_6, ordinal_cols2),
])
preprocessing
```

# Machine Learning

```python
## build ML by using all models with thier default perameters to determine the best model with best score:

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import  RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.model_selection import cross_validate

models = [
    ('Logistic Regression', LogisticRegression(random_state= 42)),
    ('KNN Classifier', KNeighborsClassifier()),
    # ('SVC', SVC(random_state= 42)),
    ('Gaussian NB', GaussianNB()),
    ('Decision Tree Classifier', DecisionTreeClassifier(random_state= 42)),
    ('Random Forest Classifier', RandomForestClassifier(random_state= 42)),
    ('XGB Classifier', XGBClassifier(random_state= 42))
]

for model in models:

    model_pipeline = Pipeline(steps= [('Preprocessing', preprocessing), ('Model', model[1])])

    cv_result = cross_validate(model_pipeline, x, y, scoring= 'accuracy', cv= 5, return_train_score= True, n_jobs= -1)

    print(model[0])

    print(f"Train Accuracy: {cv_result['train_score'].mean().round(4) * 100:.2f}%")
    print(f"Test Accuracy: {cv_result['test_score'].mean().round(4) * 100:.2f}%")

    print('-' * 50)
```

```python
## The Target is (not an extreme imbalance) , so i used another metric

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import  RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.model_selection import cross_validate

models = [
    ('Logistic Regression', LogisticRegression(random_state= 42)),
    ('KNN Classifier', KNeighborsClassifier()),
    # ('SVC', SVC(random_state= 42)),
    ('Gaussian NB', GaussianNB()),
    ('Decision Tree Classifier', DecisionTreeClassifier(random_state= 42)),
    ('Random Forest Classifier', RandomForestClassifier(random_state= 42)),
    ('XGB Classifier', XGBClassifier(random_state= 42))
]

for model in models:

    model_pipeline = Pipeline(steps= [('Preprocessing', preprocessing), ('Model', model[1])])

    cv_result = cross_validate(model_pipeline, x, y, scoring= ['accuracy', 'precision', 'recall', 'f1'], cv= 5, return_train_score= True, n_jobs= -2)

    print(model[0])

    # Accuracy Score
    print(f"Train Accuracy: {cv_result['train_accuracy'].mean().round(4) * 100:.2f}%")
    print(f"Test Accuracy: {cv_result['test_accuracy'].mean().round(4) * 100:.2f}%")

    # Precision Score
    print(f"Train Precision: {cv_result['train_precision'].mean().round(4) * 100:.2f}%")
    print(f"Test Precision: {cv_result['test_precision'].mean().round(4) * 100:.2f}%")

    # Recall Score
    print(f"Train Recall: {cv_result['train_recall'].mean().round(4) * 100:.2f}%")
    print(f"Test Recall: {cv_result['test_recall'].mean().round(4) * 100:.2f}%")

    # F1 Score
    print(f"Train F1 Score: {cv_result['train_f1'].mean().round(4) * 100:.2f}%")
    print(f"Test F1 Score: {cv_result['test_f1'].mean().round(4) * 100:.2f}%")

    print('-' * 50)
```

### Hyperparameter Tuning

- Logistic Regression

```python
from sklearn.model_selection import GridSearchCV

lr_pipeline = Pipeline(steps= [('Preprocessing', preprocessing), ('Model', LogisticRegression(random_state= 42))])

param_grid = {
    'Model__C' : [0.001, 0.01, 0.1, 1, 10, 100],
    'Model__penalty': ['l2' ]
}

grid_search_lr = GridSearchCV(estimator= lr_pipeline, param_grid= param_grid, scoring= 'f1', cv= 5, n_jobs= -2, return_train_score= True)
grid_search_lr.fit(x, y)
```

```python
from sklearn.model_selection import GridSearchCV

xgb_pipeline = Pipeline(steps= [('Preprocessing', preprocessing), ('Model', XGBClassifier(random_state= 42))])

param_grid = {
    'Model__n_estimators': [200, 300, 400],
    'Model__max_depth': [3, 4, 5],
    'Model__learning_rate': [0.01, 0.03,  0.1]
}

grid_search_xgb = GridSearchCV(estimator= xgb_pipeline, param_grid= param_grid, scoring= 'f1', cv= 5, n_jobs= -2, return_train_score= True)

grid_search_xgb.fit(x, y)
```
